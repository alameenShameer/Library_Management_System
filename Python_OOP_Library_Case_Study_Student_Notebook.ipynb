## The scenario
A college library keeps its books and members in Python **dictionaries**. Staff write the issue and return details by hand and often make mistakes: they issue books that are out of stock, or let a student take more books than allowed. **You will turn the library into objects**: books know how many copies they have, students and teachers know how many books they may borrow, and the library object connects them.

## What you will practise
* **Classes and objects:** `__init__`, `self`, attributes and methods
* `__str__` to print objects nicely
* **Encapsulation:** a private attribute (`__copies`) changed only through methods
* **Inheritance:** `Student` and `Teacher` inherit from `Member` and override its methods
* **Polymorphism:** the same method name (`max_books()`) gives a different answer for each class
* Dictionaries and lists of objects; creating objects from dictionary data
